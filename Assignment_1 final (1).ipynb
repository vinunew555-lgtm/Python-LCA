```python
mylist= [ " physics", "chemistry", "mathematics", "python learning"]
print ("1.the initial list is:", mylist)
#access list items 
print("2.the 2nd element in the list is:",mylist[2])

#change list items
mylist[2]="yoga"
print("3.after adding yoga to the list,the new second element in the list is:", mylist[2])

#add list items-method 1
mylist.append("university core")
print ("4.after adding university code to the last of the list, the new list is:", mylist)

#add list items-method 2
mylist.insert(3, "ALDC")
print("5.after adding ALDC to 3rd and Sports to the fourth position, the new list is:", mylist)

#remove list items- method 1
mylist.remove("ALDC")
print("6.after removing ALDC from the list, the new list is:", mylist)

#remove list items- method 2
mylist.pop(4)
print("7.after removing sports, the new list is:", mylist)

#sort list alphanumerically
mylist.sort
print("8.after sorting the list, the new ordered list is:", mylist)

#copy list
copylist=mylist.copy()
print("9.the copied list of mylist is:", copylist)

#joining two lists
list2=[1,2,3]
mylist= [ " physics", "chemistry", "mathematics", "python learning"]
print("if we wish to combine 2 lists, for example if i join [list2=1,2,3] to mylist , it would look like following:")
list3=list2+mylist
print("the list after combining list 2 and mylist is:",list3)

```

```python
#tuples
mytuple=("red", "white" ,"blue", "purple", "green")
print("1. the initial tuple is:", mytuple)

#accessing tuple items.
print("2.same as the list, the 3rd item of the tuple is:", mytuple[3])

#negative indexing
print("3.for refering to the last of the tuples, we can use negative indexing. -1 refers to the last item and so on")
print("the last item of the tuple is:", mytuple[-1])

#range of indexes
print("4.When specifying a range, the return value will be a new tuple with the specified items.")
print("the items within the range of 1 to 3 is:", mytuple[1:3])

#updating tuples
print("5.Tuples are unchangeable, or immutable,but you can convert the tuple into a list, change the list, and convert the list back into a tuple.")
mytuple=("red", "white" ,"blue", "purple", "green")
v=list(mytuple)
v[0]="pink"
mylist=tuple(v)
print("6.when we updated the list and changed the first element, we get the final list as:", mylist)

#unpacking tuples
print("7.Tuple unpacking is used to take the individual values stored inside a tuple and assign them to separate variables in one line.")
(apple, milk, sky, purple, grass)=mytuple
print("so if we assign red to apple, green to grass and so on, we get these outputs by printing the assigned variables, like:",apple,milk,sky,grass)

#joining tuples
tuple1=("a","b","c")
tuple2=("1","2","3")

tuple3= tuple1 + tuple2

print("8.if we join tuple 1 and tuple 2 containing a,b,c and 1,2,3 respectively, we get the final joined tuple 3 as:", tuple3)

#functions in a tuple
#count()	Returns the number of times a specified value occurs in a tuple
#index()	Searches the tuple for a specified value and returns the position of where it was found
```

```python
student = {
    "name": "Vinayak",
    "age": 19,
    "course": "Python",
    "city": "Solapur",
    "marks": 97,
}


print("1. Displaying the student's name")
print("Student name:", student["name"])


print("2. Displaying the student's course")
print("Student course:", student["course"])


print("3. Checking if 'age' is present in the dictionary")
print("If 'age' is present in the dictionary, the code will display 'Age is present in the dictionary'")
if "age" in student:
    print("Age is present in the dictionary")
else:
    print("Age not found")


print("4. Changing the student's marks")
student["marks"] = 90
print("Updated marks:", student["marks"])


print("5. Finding the total number of items")
print("Total items:", len(student))


print("6. Adding a new item")
student["email"] = "Vinayak@example.com"
print("Updated student dictionary:", student)


print("7. Removing the city from the dictionary")
student.pop("city")
print("Dictionary after removing city:", student)


print("8. Displaying all the keys")
print("The keys are:", student.keys())


print("9. Displaying all the values")
print("The values are:", student.values())


print("10. Creating a backup")
backup_student = student.copy()


print("11. Creating another student's dictionary")
friend_student = {
    "name": "Amit",
    "age": 21,
    "course": "Data Science",
    "marks": 88
}


print("12. Displaying final dictionaries")
print("Final student dictionary:", student)
print("Backup dictionary:", backup_student)
print("Friend's dictionary:", friend_student)


# DICTIONARY
# A dictionary stores data in KEY : VALUE pairs.
# Example: "name" is the KEY and "Vinayak" is the VALUE.

# FEATURES OF A DICTIONARY:
# 1. Dictionaries store data as key : value pairs.
# 2. Dictionaries are changeable (mutable).
# 3. Dictionaries are ordered in modern Python (Python 3.7+).
# 4. Dictionary keys must be UNIQUE.
# 5. Values can be duplicated.
# 6. Values can be of any data type.
# 7. Keys should be immutable/hashable types such as str, int, float or tuple.
# 8. Dictionaries can contain different data types.
# 9. Dictionaries can be nested inside other dictionaries.
#
# IMPORTANT LIMITATIONS / RULES:
# 1. You CANNOT have two different values with the same key.
#    If you use the same key again, the old value gets replaced.
#
#    Example:
#    student = {"name": "Vinayak", "name": "Amit"}
#    Result -> {"name": "Amit"}
#
# 2. Keys must be unique.
# 3. A list cannot normally be used as a dictionary key because lists are mutable.
#
#    Wrong:
#    student = {[1, 2]: "numbers"}
#
#    Correct:
#    student = {(1, 2): "numbers"}
#
# 4. You cannot access dictionary items using a numerical index like a list.
#
#    Wrong:
#    student[0]
#
#    Correct:
#    student["name"]
#
# 5. Trying to access a key that does not exist using [] gives a KeyError.
#
#    Wrong if "phone" doesn't exist:
#    print(student["phone"])
#
#    Safer:
#    print(student.get("phone"))
#
# 6. Dictionaries are not suitable when you need duplicate keys.
# 7. Dictionary keys must be hashable, but values can be almost any Python object.
```
